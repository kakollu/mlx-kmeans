```python
from pathlib import Path
from mlx_kmeans import cluster_file

# Replace this with your CSV, TSV or XLSX path.
file = Path('customers.csv')
columns = ['annual_spend', 'orders', 'visits']
k = 3
n_init = 10  # similar purpose to nstart in R
```

```python
result = cluster_file(file, columns=columns, k=k, n_init=n_init, seed=0)
result['summary']
```

Features are standardized by default. Summary means are in original units. Each input row is exported with a 1-based cluster number. Numbers identify groups; they do not rank customers.
Missing or nonnumeric selected values raise an error. Clean the data, or explicitly choose `missing='drop'`; excluded rows remain in the export with blank labels.
This uses Lloyd's algorithm and a relative-inertia stopping rule. R's default algorithm and initialization differ, so do not expect identical clusters. Try several values of k and interpret the group sizes and feature means.

```python
print('Results:', result['output'])
print('Rows clustered:', result['run']['clustered_rows'])
print('Rows excluded:', result['run']['excluded_rows'])
print('Inertia in the fitted feature space:', result['run']['inertia'])
```
